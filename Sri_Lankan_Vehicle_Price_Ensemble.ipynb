```python

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Machine learning libraries imported!")
```

```python

# Prepare a clean copy
model_df = df_clean.copy()

# Remove records with invalid target values
model_df = model_df[
    model_df['Price'].notna() &
    (model_df['Price'] > 0)
].copy()

# Define numerical features
numerical_features = [
    'YOM',
    'Engine (cc)',
    'Millage(KM)',
    'Vehicle_Age'
]

# Define categorical features
categorical_features = [
    'Brand',
    'Model',
    'Gear',
    'Fuel Type',
    'Town',
    'Leasing',
    'Condition',
    'AIR CONDITION',
    'POWER STEERING',
    'POWER MIRROR',
    'POWER WINDOW'
]

# Input features
X = model_df[
    numerical_features + categorical_features
]

# Target variable
y = model_df['Price']

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python

# Numerical preprocessing
numeric_transformer = StandardScaler()

# Categorical preprocessing
categorical_transformer = OneHotEncoder(
    handle_unknown='ignore',
    min_frequency=5,
    sparse_output=True
)

# Combine preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline ready!")
```

```python

# Create Random Forest Regressor
rf_model = RandomForestRegressor(
    n_estimators=300,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features=0.8,
    random_state=42,
    n_jobs=-1
)

# Create complete pipeline
rf_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', rf_model)
    ]
)

print("Random Forest model created!")
```

```python

print("Training Random Forest Model...")

rf_pipeline.fit(X_train, y_train)

print("Random Forest training completed!")
```

```python

# Predict testing data
rf_predictions = rf_pipeline.predict(X_test)

# Compare actual and predicted values
results = pd.DataFrame({
    'Actual Price': y_test.values,
    'Predicted Price': rf_predictions
})

display(results.head(10))
```

```python

# Calculate evaluation metrics
rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

rf_rmse = np.sqrt(
    mean_squared_error(y_test, rf_predictions)
)

rf_r2 = r2_score(
    y_test,
    rf_predictions
)

print("RANDOM FOREST PERFORMANCE")
print("--------------------------")
print("MAE :", round(rf_mae, 4))
print("RMSE:", round(rf_rmse, 4))
print("R2 Score:", round(rf_r2, 4))
```

```python

plt.figure(figsize=(9, 6))

plt.scatter(
    y_test,
    rf_predictions,
    alpha=0.5
)

# Perfect prediction line
minimum = min(y_test.min(), rf_predictions.min())
maximum = max(y_test.max(), rf_predictions.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    'r--'
)

plt.title('Random Forest: Actual vs Predicted Prices')
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')

plt.show()
```

```python

# Get feature names after preprocessing
feature_names = (
    rf_pipeline
    .named_steps['preprocessor']
    .get_feature_names_out()
)

# Get feature importance
importance = (
    rf_pipeline
    .named_steps['model']
    .feature_importances_
)

# Create importance dataframe
importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importance
})

# Top 15 important features
importance_df = importance_df.sort_values(
    by='Importance',
    ascending=False
).head(15)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=importance_df,
    x='Importance',
    y='Feature'
)

plt.title('Top 15 Random Forest Feature Importances')
plt.xlabel('Importance')
plt.ylabel('Feature')

plt.show()
```

```python

from xgboost import XGBRegressor
from catboost import CatBoostRegressor
from sklearn.ensemble import GradientBoostingRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.pipeline import Pipeline
import numpy as np
import pandas as pd

print("Boosting libraries imported successfully!")
```

```python

def evaluate_model(model, X_test, y_test, model_name):

    predictions = model.predict(X_test)

    mae = mean_absolute_error(
        y_test, predictions
    )

    rmse = np.sqrt(
        mean_squared_error(y_test, predictions)
    )

    r2 = r2_score(
        y_test, predictions
    )

    print(f"\n{model_name} Performance")
    print("-----------------------------")
    print("MAE :", round(mae, 4))
    print("RMSE:", round(rmse, 4))
    print("R2 Score:", round(r2, 4))

    return {
        'Model': model_name,
        'MAE': mae,
        'RMSE': rmse,
        'R2 Score': r2
    }
```

```python

xgb_model = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1,
    objective='reg:squarederror'
)

xgb_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', xgb_model)
    ]
)

print("XGBoost model created!")
```

```python

print("Training XGBoost...")

xgb_pipeline.fit(
    X_train,
    y_train
)

print("XGBoost training completed!")
```

```python

xgb_result = evaluate_model(
    xgb_pipeline,
    X_test,
    y_test,
    "XGBoost"
)
```

```python

cat_model = CatBoostRegressor(
    iterations=300,
    learning_rate=0.05,
    depth=6,
    loss_function='RMSE',
    random_seed=42,
    verbose=0,
    allow_writing_files=False
)

cat_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', cat_model)
    ]
)

print("CatBoost model created!")
```

```python

print("Training CatBoost...")

cat_pipeline.fit(
    X_train,
    y_train
)

print("CatBoost training completed!")
```

```python

cat_result = evaluate_model(
    cat_pipeline,
    X_test,
    y_test,
    "CatBoost"
)
```

```python

gb_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    min_samples_split=5,
    random_state=42
)

gb_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', gb_model)
    ]
)

print("Gradient Boosting model created!")
```

```python

print("Training Gradient Boosting...")

gb_pipeline.fit(
    X_train,
    y_train
)

print("Gradient Boosting training completed!")
```

```python

from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import Ridge

from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
import pandas as pd
import joblib

print("Stacking libraries imported successfully!")
```

```python

# Base Model 1: Random Forest
rf_base = Pipeline([
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(
        n_estimators=200,
        max_depth=None,
        min_samples_leaf=1,
        random_state=42,
        n_jobs=1
    ))
])

# Base Model 2: XGBoost
xgb_base = Pipeline([
    ('preprocessor', preprocessor),
    ('model', XGBRegressor(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        objective='reg:squarederror',
        random_state=42,
        n_jobs=1
    ))
])

# Base Model 3: CatBoost
cat_base = Pipeline([
    ('preprocessor', preprocessor),
    ('model', CatBoostRegressor(
        iterations=200,
        learning_rate=0.05,
        depth=6,
        loss_function='RMSE',
        random_seed=42,
        verbose=0,
        allow_writing_files=False,
        thread_count=1
    ))
])

print("Three base models created successfully!")
```

```python

meta_learner = Ridge(alpha=1.0)

print("Ridge Regression Meta-Learner created!")
```

```python

stacking_model = StackingRegressor(
    estimators=[
        ('random_forest', rf_base),
        ('xgboost', xgb_base),
        ('catboost', cat_base)
    ],
    final_estimator=meta_learner,
    cv=5,
    n_jobs=-1,
    passthrough=False
)

print("Stacking Regressor created successfully!")
```

```python

print("Training Stacking Ensemble...")

stacking_model.fit(
    X_train,
    y_train
)

print("Stacking model training completed!")
```

```python

stacking_predictions = stacking_model.predict(X_test)

stacking_results = pd.DataFrame({
    'Actual Price': y_test.values,
    'Predicted Price': stacking_predictions
})

display(stacking_results.head(10))
```

```python

stacking_mae = mean_absolute_error(
    y_test,
    stacking_predictions
)

stacking_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        stacking_predictions
    )
)

stacking_r2 = r2_score(
    y_test,
    stacking_predictions
)

print("STACKING MODEL PERFORMANCE")
print("---------------------------")
print("MAE:", round(stacking_mae, 4))
print("RMSE:", round(stacking_rmse, 4))
print("R2 Score:", round(stacking_r2, 4))
```

```python

ann_regressor = MLPRegressor(
    hidden_layer_sizes=(128, 64),
    activation='relu',
    solver='adam',
    alpha=0.001,
    learning_rate_init=0.001,
    max_iter=300,
    early_stopping=True,
    validation_fraction=0.1,
    n_iter_no_change=20,
    random_state=42
)

# Build ANN pipeline
ann_base_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', ann_regressor)
    ]
)

# Scale the target variable during training
ann_pipeline = TransformedTargetRegressor(
    regressor=ann_base_pipeline,
    transformer=StandardScaler()
)

print("ANN model created successfully!")
```

## Model Performance Comparison and Interpretation

Six machine learning models were implemented and evaluated for Sri Lankan vehicle price prediction.

The Random Forest and RF + ANN Hybrid models achieved the best overall performance.

The RF + ANN Hybrid model achieved the lowest RMSE of 12.8784 and the highest R² score of 0.9076. This indicates that the hybrid model explained approximately 90.76% of the variation in vehicle prices in the test dataset.

Random Forest achieved the lowest MAE of 4.9239, indicating that it produced the lowest average absolute prediction error among the evaluated models.

The Stacking model achieved an R² score of 0.9009, showing that combining Random Forest, XGBoost and CatBoost predictions also produced a strong result.

Among the boosting models, XGBoost performed best, with an RMSE of 13.8912 and an R² score of 0.8925. CatBoost achieved an R² score of 0.8786, while Gradient Boosting achieved 0.8545.

The results show that ensemble techniques can produce different prediction performances depending on the dataset and model configuration. The hybrid model achieved the best RMSE, but it did not outperform Random Forest in MAE. Therefore, the best model depends on the evaluation metric being considered.
